```python
#!/bin/bash
# Download the book Alice's Adventures in Wonderland. The saved file name will be book.txt
wget -O book.txt https://www.gutenberg.org/cache/epub/11/pg11.txt

# Check the list of available files in the current directory
ls

# Rename the book from book.txt to Alice.txt
mv book.txt  Alice.txt

# Find out the 10 most frequent words and their frequencies
echo The 10 most frequent words and their frequencies
cat Alice.txt | #Start working on the file by reading or showing it by “cat”
tr ' ' '\n'|  # Put each word in a separate line
tr '[:upper:]' '[:lower:]'| #Convert all alphabets into lowercase
tr '[:punct:]' ' '| #Remove punctuations or transform them into space
tr '^”$' ' '|  #I found some extra quotation marks near words which I was not able to remove by the previous command, and adding + to the end
grep -v '^$'| # remove empty lines
sort | # Sort words alphabetically
uniq -c | # Count the occurrences of each unique word
sort -nr| #Sort the results from the highest to the lowest numerically
head #Show the top 10 words and occurrences

#Find out how many lines have the word ”and” (or ”ja” in Finnish)
echo How many lines have the word \“and\”
grep -iwc 'and' Alice.txt #count the lines which match while -i ignore cases and -w matches just words without being part of another word

#Replace all upper case letters with lower case letters and save the result as a new file
echo Transform uppercases to lowercases and save in a new file
cat Alice.txt| tr '[:upper:]' '[:lower:]' > Alice_LowerCases.txt

#Exclude metadata -> The actual books located between two structures in the Gothenburg project, which are used here to make a new file having only the book text.
#The -n option means don't print by default; then the script says to print between the lines mentioned as follows.
cat Alice.txt | sed -n '/START OF THE PROJECT GUTENBERG EBOOK/, /END OF THE PROJECT GUTENBERG EBOOK/p' > Alice_clean.txt
```

```python
%%bash

# Download the book Alice's Adventures in Wonderland.
wget -q -O book.txt https://www.gutenberg.org/cache/epub/11/pg11.txt

# Check the list of available files in the current directory
ls

# Rename the book from book.txt to Alice.txt
mv book.txt Alice.txt

# Find out the 10 most frequent words and their frequencies
echo "--- The 10 most frequent words and their frequencies ---"
cat Alice.txt | \
tr ' ' '\n' | \
tr '[:upper:]' '[:lower:]' | \
tr '[:punct:]' ' ' | \
tr '^”$' ' ' | \
grep -v '^$' | \
sort | \
uniq -c | \
sort -nr | \
head

# Find out how many lines have the word "and"
echo -e "\n--- How many lines have the word 'and' ---"
grep -iwc 'and' Alice.txt

# Replace all upper case letters with lower case letters and save the result as a new file
echo -e "\n--- Transform uppercases to lowercases and save in a new file ---"
cat Alice.txt | tr '[:upper:]' '[:lower:]' > Alice_LowerCases.txt

# Exclude metadata and create a clean file
echo -e "\n--- Extracting book text to Alice_clean.txt ---"
cat Alice.txt | sed -n '/START OF THE PROJECT GUTENBERG EBOOK/, /END OF THE PROJECT GUTENBERG EBOOK/p' > Alice_clean.txt

# Verify the new files exist
ls -lh Alice_clean.txt Alice_LowerCases.txt
```
